# Fetal Health Prediction using Machine Learning

## Notebook 2 : Exploratory Data Analysis (EDA)

### Objective

The objective of this notebook is to explore the fetal health dataset using statistical summaries and visualizations to identify patterns, relationships, and class distributions before model training.

### Tasks

- Import visualization libraries
- Load cleaned dataset
- Analyze target class distribution
- Visualize feature distributions
- Plot correlation heatmap
- Detect outliers
- Summarize observations

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../01_dataset/01_raw/fetal_health.csv")
```

```python
# Check target distribution

print(df["fetal_health"].value_counts())

sns.countplot(data=df, x="fetal_health")
plt.title("Fetal Health Class Distribution")
plt.xlabel("Fetal Health Class")
plt.ylabel("Count")
plt.show()
```

```python
#Target percentage
class_percentage = df["fetal_health"].value_counts(normalize=True) * 100
print(class_percentage)
```

```python
#Feature distributions
df.hist(figsize=(20, 15), bins=20)
plt.tight_layout()
plt.show()
```

```python
#Correlation heatmap
plt.figure(figsize=(18, 14))

sns.heatmap(
    df.corr(),
    cmap="coolwarm",
    annot=False
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
#Correlation with target
target_corr = df.corr()["fetal_health"].sort_values(ascending=False)

print(target_corr)
```

```python
#Top correlated features
target_corr.drop("fetal_health").abs().sort_values(ascending=False).head(10)
```

```python
#Boxplot
plt.figure(figsize=(20, 12))

sns.boxplot(data=df)

plt.xticks(rotation=90)
plt.title("Feature Boxplots")
plt.tight_layout()
plt.show()
```
